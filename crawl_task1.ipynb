```python
# 동적 크롤링을 위한 설정
def dynamic_crawl(row):
    driver = webdriver.Chrome()
    driver.set_page_load_timeout(60)
    try:
      driver.get(row['URL'])
    except TimeoutException:
      print(f"페이지 로딩 시간이 초과되었습니다: {row['SITE_NAME']}")
    time.sleep(10)
    page_source = driver.page_source
    soup = BeautifulSoup(page_source, 'html.parser')
    driver.quit()

    return soup

def dynamic_crawl_1(row):
    driver = webdriver.Chrome()
    driver.get(row['URL'])
    time.sleep(10)
    button = driver.find_element(By.ID, 'ofr_pageSize')  # 버튼의 ID로 요소 찾기
    button.click()  # 클릭
    button = driver.find_element(By.XPATH, '//*[@id="ofr_pageSize"]/option[1]')  # XPath로 요소 찾기
    button.click()  # 클릭
    page_source = driver.page_source
    soup = BeautifulSoup(page_source, 'html.parser')
    driver.quit()
    return soup

def dynamic_crawl_2(row):
    driver = webdriver.Chrome()
    driver.get(row['URL'])
    time.sleep(10)
    reset_button = driver.find_element(By.CSS_SELECTOR, row['click_button'])
    reset_button.click()
    page_source = driver.page_source
    soup = BeautifulSoup(page_source, 'html.parser')
    driver.quit()
    return soup

def click_dynamic_crawl(row, page_number):
    driver = webdriver.Chrome()
    driver.get(row['URL'])
    time.sleep(5)
    reset_button = driver.find_element(By.CSS_SELECTOR, f"body > form > div.default_board > div.paging > table > tbody > tr > td:nth-child(4) > span:nth-child({(page_number-1)*2+1}) > a")
    reset_button.click()
    print(f"body > form > div.default_board > div.paging > table > tbody > tr > td:nth-child(4) > span:nth-child({(page_number-1)*2+1}) > a")
    time.sleep(5)
    page_source = driver.page_source
    soup = BeautifulSoup(page_source, 'html.parser')
    driver.quit()
    return soup

def click_dynamic_crawl1(row, page_number):
    driver = webdriver.Chrome()
    driver.get(row['URL'])
    time.sleep(5)
    reset_button = driver.find_element(By.CSS_SELECTOR, f"#form1 > div.pgeAbs.mt30 > p > span:nth-child({(page_number)}) > a")
    reset_button.click()
    print(f"#form1 > div.pgeAbs.mt30 > p > span:nth-child({(page_number)}) > a")
    time.sleep(5)
    page_source = driver.page_source
    soup = BeautifulSoup(page_source, 'html.parser')
    driver.quit()
    return soup



def click_dynamic_crawl2(row, page_number):
    driver = webdriver.Chrome()
    driver.get(row['URL'])
    time.sleep(5)
    reset_button = driver.find_element(By.XPATH, f"/html/body/div[2]/div[2]/div/section[2]/div[1]/form/div[2]/a[{(page_number)+2}]")
    #element = driver.find_element(By.CSS_SELECTOR, f"#listForm > div.box_page > a:nth-child({(page_number)+2})")
    #driver.execute_script("arguments[0].scrollIntoView();", element)
    #element.click()
    reset_button.click()
    print(f"#listForm > div.box_page > a:nth-child({(page_number)+2})")
    time.sleep(5)
    page_source = driver.page_source
    soup = BeautifulSoup(page_source, 'html.parser')
    driver.quit()
    return soup

def click_dynamic_crawl3(row, page_number):
    driver = webdriver.Chrome()
    driver.get(row['URL'])
    time.sleep(5)
    reset_button = driver.find_element(By.CSS_SELECTOR, f"#txt > div.text-center > div > ul > li:nth-child({(page_number)+2}) > a")
    reset_button.click()
    print(f"#txt > div.text-center > div > ul > li:nth-child({(page_number)+2}) > a")
    time.sleep(5)
    page_source = driver.page_source
    soup = BeautifulSoup(page_source, 'html.parser')
    driver.quit()
    return soup

def click_dynamic_crawl4(row, page_number):
    driver = webdriver.Chrome()
    driver.get(row['URL'])
    time.sleep(5)
    reset_button = driver.find_element(By.CSS_SELECTOR, f"#dataForm > div.pagination.mt-md-4 > a:nth-child({(page_number)})")
    reset_button.click()
    print(f"#dataForm > div.pagination.mt-md-4 > a:nth-child({(page_number)})")
    time.sleep(5)
    page_source = driver.page_source
    soup = BeautifulSoup(page_source, 'html.parser')
    driver.quit()
    return soup

def click_dynamic_crawl5(row, page_number):
    driver = webdriver.Chrome()
    driver.get(row['URL'])
    time.sleep(5)
    reset_button = driver.find_element(By.CSS_SELECTOR, f"#cont-body > div.paging > div > div > a:nth-child({(page_number+2)})")
    reset_button.click()
    print(f"#cont-body > div.paging > div > div > a:nth-child({(page_number+2)})")
    time.sleep(5)
    page_source = driver.page_source
    soup = BeautifulSoup(page_source, 'html.parser')
    driver.quit()
    return soup

def click_dynamic_crawl6(row, page_number):
    driver = webdriver.Chrome()
    driver.get(row['URL'])
    time.sleep(5)
    reset_button = driver.find_element(By.CSS_SELECTOR, f"#contentDiv > form > table.MAT10 > tbody > tr > td > table > tbody > tr > td > table > tbody > tr > td:nth-child(4) > span:nth-child({(page_number-1)*2+1}) > a")
    reset_button.click()
    print(f"#contentDiv > form > table.MAT10 > tbody > tr > td > table > tbody > tr > td > table > tbody > tr > td:nth-child(4) > span:nth-child({(page_number-1)*2+1}) > a")
    time.sleep(5)
    page_source = driver.page_source
    soup = BeautifulSoup(page_source, 'html.parser')
    driver.quit()
    return soup

def click_dynamic_crawl7(row, page_number):
    driver = webdriver.Chrome()
    driver.get(row['URL'])
    time.sleep(5)
    reset_button = driver.find_element(By.CSS_SELECTOR, f"#list > div.bod_page > a:nth-child({(page_number)+2})")
    reset_button.click()
    print(f"#list > div.bod_page > a:nth-child({(page_number)+2})")
    time.sleep(5)
    page_source = driver.page_source
    soup = BeautifulSoup(page_source, 'html.parser')
    driver.quit()
    return soup

def click_dynamic_crawl8(row, page_number):
    driver = webdriver.Chrome()
    driver.get(row['URL'])
    time.sleep(5)
    reset_button = driver.find_element(By.CSS_SELECTOR, f"#board > div:nth-child(4) > table > tbody > tr > td > table > tbody > tr > td:nth-child(4) > span:nth-child({(page_number-1)*2+1}) > a")
    reset_button.click()
    print(f"#board > div:nth-child(4) > table > tbody > tr > td > table > tbody > tr > td:nth-child(4) > span:nth-child({(page_number-1)*2+1}) > a")
    time.sleep(5)
    page_source = driver.page_source
    soup = BeautifulSoup(page_source, 'html.parser')
    driver.quit()
    return soup

def click_dynamic_crawl9(row, page_number):
    driver = webdriver.Chrome()
    driver.get(row['URL'])
    time.sleep(9)
    reset_button = driver.find_element(By.CSS_SELECTOR, f"body > form > div.sb_w > div:nth-child(3) > table > tbody > tr > td > table > tbody > tr > td:nth-child(4) > span:nth-child({(page_number-1)*2+1}) > a")
    reset_button.click()
    print(f"body > form > div.sb_w > div:nth-child(3) > table > tbody > tr > td > table > tbody > tr > td:nth-child(4) > span:nth-child({(page_number-1)*2+1}) > a")
    time.sleep(5)
    page_source = driver.page_source
    soup = BeautifulSoup(page_source, 'html.parser')
    driver.quit()
    return soup

def click_dynamic_crawl10(row, page_number):
    driver = webdriver.Chrome()
    driver.set_page_load_timeout(10)
    driver.get(row['URL'])
    time.sleep(7)
    reset_button = driver.find_element(By.XPATH, f"/html/body/form/table[3]/tbody/tr/td/table[2]/tbody/tr/td/table/tbody/tr/td/table/tbody/tr/td[4]/span[{(page_number-1)*2+1}]/a")
    reset_button.click()
    print(f"/html/body/form/table[3]/tbody/tr/td/table[2]/tbody/tr/td/table/tbody/tr/td/table/tbody/tr/td[4]/span[{(page_number-1)*2+1}]/a")
    time.sleep(7)
    page_source = driver.page_source
    soup = BeautifulSoup(page_source, 'html.parser')
    driver.quit()
    return soup

def click_dynamic_crawl11(row, page_number):
    driver = webdriver.Chrome()
    driver.set_page_load_timeout(10)
    driver.get(row['URL'])
    time.sleep(7)
    reset_button = driver.find_element(By.XPATH, f"//*[@id='list']/div[2]/div/a[{page_number+2}]")
    reset_button.click()
    print(f"//*[@id='list']/div[2]/div/a[{page_number+2}]")
    time.sleep(7)
    page_source = driver.page_source
    soup = BeautifulSoup(page_source, 'html.parser')
    driver.quit()
    return soup

def click_dynamic_crawl12(row, page_number):
    driver = webdriver.Chrome()
    driver.set_page_load_timeout(10)
    driver.get(row['URL'])
    time.sleep(7)
    reset_button = driver.find_element(By.CSS_SELECTOR, f"#sidoGosiAPIVO > div.pagination > div.normal_pagination > a:nth-child({(page_number)+2})")
    #reset_button = driver.find_element(By.XPATH, f"//*[@id='list']/div[2]/div/a[{page_number+2}]")
    reset_button.click()
    print(f"#sidoGosiAPIVO > div.pagination > div.normal_pagination > a:nth-child({(page_number)+2})")
    time.sleep(7)
    page_source = driver.page_source
    soup = BeautifulSoup(page_source, 'html.parser')
    driver.quit()
    return soup

def click_dynamic_crawl13(row, page_number):
    driver = webdriver.Chrome()
    driver.set_page_load_timeout(10)
    driver.get(row['URL'])
    time.sleep(7)
    reset_button = driver.find_element(By.CSS_SELECTOR, f"#txt > div > div.text-center > ul > li:nth-child({(page_number)+2}) > a")
    #reset_button = driver.find_element(By.XPATH, f"//*[@id='list']/div[2]/div/a[{page_number+2}]")
    reset_button.click()
    print(f"#txt > div > div.text-center > ul > li:nth-child({(page_number)+2}) > a")
    time.sleep(7)
    page_source = driver.page_source
    soup = BeautifulSoup(page_source, 'html.parser')
    driver.quit()
    return soup

def click_dynamic_crawl14(row, page_number):
    driver = webdriver.Chrome()
    driver.set_page_load_timeout(10)
    driver.get(row['URL'])
    time.sleep(7)
    reset_button = driver.find_element(By.CSS_SELECTOR, f"body > form > div > div > div.p-pagination > div > span.p-page__link-group > a:nth-child({(page_number)})")
    #reset_button = driver.find_element(By.XPATH, f"//*[@id='list']/div[2]/div/a[{page_number+2}]")
    reset_button.click()
    print(f"body > form > div > div > div.p-pagination > div > span.p-page__link-group > a:nth-child({(page_number)})")
    time.sleep(7)
    page_source = driver.page_source
    soup = BeautifulSoup(page_source, 'html.parser')
    driver.quit()
    return soup

def click_dynamic_crawl15(row, page_number):
    driver = webdriver.Chrome()
    driver.set_page_load_timeout(10)
    driver.get(row['URL'])
    time.sleep(7)
    reset_button = driver.find_element(By.CSS_SELECTOR, f"body > form > div > div.paging > table > tbody > tr > td:nth-child(4) > span:nth-child({(page_number-1)*2+1}) > a")
    #reset_button = driver.find_element(By.XPATH, f"//*[@id='list']/div[2]/div/a[{page_number+2}]")
    reset_button.click()
    print(f"body > form > div > div.paging > table > tbody > tr > td:nth-child(4) > span:nth-child({(page_number-1)*2+1}) > a")
    time.sleep(7)
    page_source = driver.page_source
    soup = BeautifulSoup(page_source, 'html.parser')
    driver.quit()
    return soup

def click_dynamic_crawl16(row, page_number):
    driver = webdriver.Chrome()
    driver.set_page_load_timeout(10)
    driver.get(row['URL'])
    time.sleep(7)
    reset_button = driver.find_element(By.CSS_SELECTOR, f"body > form > table > tbody > tr:nth-child(2) > td:nth-child(2) > table > tbody > tr:nth-child(7) > td > table > tbody > tr > td:nth-child({(page_number)+4}) > a")
    #reset_button = driver.find_element(By.XPATH, f"//*[@id='list']/div[2]/div/a[{page_number+2}]")
    reset_button.click()
    print(f"body > form > table > tbody > tr:nth-child(2) > td:nth-child(2) > table > tbody > tr:nth-child(7) > td > table > tbody > tr > td:nth-child({(page_number)+4}) > a")
    time.sleep(7)
    page_source = driver.page_source
    soup = BeautifulSoup(page_source, 'html.parser')
    driver.quit()
    return soup

def click_dynamic_crawl17(row, page_number):
    driver = webdriver.Chrome()
    driver.set_page_load_timeout(10)
    driver.get(row['URL'])
    time.sleep(7)
    reset_button = driver.find_element(By.CSS_SELECTOR, f"body > form > div.board > div > div > table > tbody > tr > td:nth-child(2) > table > tbody > tr > td:nth-child(4) > span:nth-child({(page_number-1)*2+1})")
    #reset_button = driver.find_element(By.XPATH, f"//*[@id='list']/div[2]/div/a[{page_number+2}]")
    reset_button.click()
    print(f"body > form > div.board > div > div > table > tbody > tr > td:nth-child(2) > table > tbody > tr > td:nth-child(4) > span:nth-child({(page_number-1)*2+1})")
    time.sleep(7)
    page_source = driver.page_source
    soup = BeautifulSoup(page_source, 'html.parser')
    driver.quit()
    return soup


def click_dynamic_crawl18(row, page_number):
    driver = webdriver.Chrome()
    driver.set_page_load_timeout(10)
    driver.get(row['URL'])
    time.sleep(7)
    reset_button = driver.find_element(By.CSS_SELECTOR, f"#contents > div > div.p-wrap.bbs.bbs_list > div.p-pagination > div.p-page_link-group > a:nth-child({page_number})")
    #reset_button = driver.find_element(By.XPATH, f"//*[@id='list']/div[2]/div/a[{page_number+2}]")
    reset_button.click()
    print(f"#contents > div > div.p-wrap.bbs.bbs_list > div.p-pagination > div.p-page_link-group > a:nth-child({page_number})")
    time.sleep(7)
    page_source = driver.page_source
    soup = BeautifulSoup(page_source, 'html.parser')
    driver.quit()
    return soup


def click_dynamic_crawl19(row, page_number):
    driver = webdriver.Chrome()
    driver.set_page_load_timeout(10)
    driver.get(row['URL'])
    time.sleep(7)
    reset_button = driver.find_element(By.CSS_SELECTOR, f"#contents > form > table:nth-child(23) > tbody > tr:nth-child(1) > td > table > tbody > tr > td > table > tbody > tr > td:nth-child(4) > span:nth-child({(page_number-1)*2+1}) > a")
    #reset_button = driver.find_element(By.XPATH, f"//*[@id='list']/div[2]/div/a[{page_number+2}]")
    reset_button.click()
    print(f"#contents > form > table:nth-child(23) > tbody > tr:nth-child(1) > td > table > tbody > tr > td > table > tbody > tr > td:nth-child(4) > span:nth-child({(page_number-1)*2+1}) > a")
    time.sleep(7)
    page_source = driver.page_source
    soup = BeautifulSoup(page_source, 'html.parser')
    driver.quit()
    return soup   

def click_dynamic_crawl20(row, page_number):
    driver = webdriver.Chrome()
    driver.set_page_load_timeout(10)
    driver.get(row['URL'])
    time.sleep(7)
    reset_button = driver.find_element(By.CSS_SELECTOR, f"body > form > div.pagination > a:nth-child({page_number})")
    #reset_button = driver.find_element(By.XPATH, f"//*[@id='list']/div[2]/div/a[{page_number+2}]")
    reset_button.click()
    print(f"body > form > div.pagination > a:nth-child({page_number})")
    time.sleep(7)
    page_source = driver.page_source
    soup = BeautifulSoup(page_source, 'html.parser')
    driver.quit()
    return soup   

def click_dynamic_crawl21(row, page_number):
    driver = webdriver.Chrome()
    driver.set_page_load_timeout(10)
    driver.get(row['URL'])
    time.sleep(7)
    reset_button = driver.find_element(By.CSS_SELECTOR, f"body > div.pagination > table > tbody > tr > td:nth-child(4) > span:nth-child({(page_number-1)*2+1}) > a")
    #reset_button = driver.find_element(By.XPATH, f"//*[@id='list']/div[2]/div/a[{page_number+2}]")
    reset_button.click()
    print(f"body > div.pagination > table > tbody > tr > td:nth-child(4) > span:nth-child({(page_number-1)*2+1}) > a")
    time.sleep(7)
    page_source = driver.page_source
    soup = BeautifulSoup(page_source, 'html.parser')
    driver.quit()
    return soup  

def click_dynamic_crawl22(row, page_number):
    driver = webdriver.Chrome()
    driver.set_page_load_timeout(10)
    driver.get(row['URL'])
    time.sleep(7)
    reset_button = driver.find_element(By.CSS_SELECTOR, f"body > div.pagination > a:nth-child({page_number})")
    #reset_button = driver.find_element(By.XPATH, f"//*[@id='list']/div[2]/div/a[{page_number+2}]")
    reset_button.click()
    print(f"body > div.pagination > a:nth-child({page_number})")
    time.sleep(7)
    page_source = driver.page_source
    soup = BeautifulSoup(page_source, 'html.parser')
    driver.quit()
    return soup  

def click_dynamic_crawl23(row, page_number):
    driver = webdriver.Chrome()
    driver.set_page_load_timeout(10)
    driver.get(row['URL'])
    time.sleep(7)
    reset_button = driver.find_element(By.XPATH, f"/html/body/div[4]/section/div/div/div[2]/div/div/div[3]/div/ul/ul/li[{page_number+2}]")
    reset_button.click()
    print(f"/html/body/div[4]/section/div/div/div[2]/div/div/div[3]/div/ul/ul/li[{page_number+2}]")
    time.sleep(7)
    page_source = driver.page_source
    soup = BeautifulSoup(page_source, 'html.parser')
    driver.quit()
    return soup 

def click_dynamic_crawl24(row, page_number):
    driver = webdriver.Chrome()
    driver.set_page_load_timeout(10)
    driver.get(row['URL'])
    time.sleep(7)
    reset_button = driver.find_element(By.CSS_SELECTOR, f"#content_area > div.container > div > div.content > div.board_list > div.paging > ul > li:nth-child({page_number}) > a")
    #reset_button = driver.find_element(By.XPATH, f"//*[@id='list']/div[2]/div/a[{page_number+2}]")
    reset_button.click()
    print(f"#content_area > div.container > div > div.content > div.board_list > div.paging > ul > li:nth-child({page_number}) > a")
    time.sleep(7)
    page_source = driver.page_source
    soup = BeautifulSoup(page_source, 'html.parser')
    driver.quit()
    return soup

def click_dynamic_crawl25(row, page_number):
    driver = webdriver.Chrome()
    driver.set_page_load_timeout(10)
    driver.get(row['URL'])
    time.sleep(7)
    reset_button = driver.find_element(By.CSS_SELECTOR, f"#eminwonWrap > div.pagination > ul > li:nth-child({page_number}) > a")
    #reset_button = driver.find_element(By.XPATH, f"//*[@id='list']/div[2]/div/a[{page_number+2}]")
    reset_button.click()
    print(f"#eminwonWrap > div.pagination > ul > li:nth-child({page_number}) > a")
    time.sleep(7)
    page_source = driver.page_source
    soup = BeautifulSoup(page_source, 'html.parser')
    driver.quit()
    return soup

def click_dynamic_crawl26(row, page_number):
    driver = webdriver.Chrome()
    driver.set_page_load_timeout(10)
    driver.get(row['URL'])
    time.sleep(7)
    reset_button = driver.find_element(By.CSS_SELECTOR, f"#listForm > div.box_page > a:nth-child({page_number+2})")
    #reset_button = driver.find_element(By.XPATH, f"//*[@id='list']/div[2]/div/a[{page_number+2}]")
    reset_button.click()
    print(f"#listForm > div.box_page > a:nth-child({page_number+2})")
    time.sleep(7)
    page_source = driver.page_source
    soup = BeautifulSoup(page_source, 'html.parser')
    driver.quit()
    return soup

def click_dynamic_crawl27(row, page_number):
    driver = webdriver.Chrome()
    driver.set_page_load_timeout(10)
    driver.get(row['URL'])
    time.sleep(7)
    reset_button = driver.find_element(By.XPATH, f"/html/body/form/table[2]/tbody/tr/td/table[2]/tbody/tr/td/table/tbody/tr/td/table/tbody/tr/td[4]/span[{(page_number-1)*2+1}]/a")
    #reset_button = driver.find_element(By.XPATH, f"//*[@id='list']/div[2]/div/a[{page_number+2}]")
    reset_button.click()
    print(f"/html/body/form/table[2]/tbody/tr/td/table[2]/tbody/tr/td/table/tbody/tr/td/table/tbody/tr/td[4]/span[{(page_number-1)*2+1}]/a")
    time.sleep(7)
    page_source = driver.page_source
    soup = BeautifulSoup(page_source, 'html.parser')
    driver.quit()
    return soup

def click_dynamic_crawl28(row, page_number):
    driver = webdriver.Chrome()
    driver.set_page_load_timeout(10)
    driver.get(row['URL'])
    time.sleep(7)
    reset_button = driver.find_element(By.CSS_SELECTOR, f"#A-Contents > div.pager > table > tbody > tr > td > table > tbody > tr > td:nth-child(4) > span:nth-child({(page_number-1)*2+1}) > a")
    #reset_button = driver.find_element(By.XPATH, f"//*[@id='list']/div[2]/div/a[{page_number+2}]")
    reset_button.click()
    print(f"#A-Contents > div.pager > table > tbody > tr > td > table > tbody > tr > td:nth-child(4) > span:nth-child({(page_number-1)*2+1}) > a")
    time.sleep(7)
    page_source = driver.page_source
    soup = BeautifulSoup(page_source, 'html.parser')
    driver.quit()
    return soup

def click_dynamic_crawl29(row, page_number):
    driver = webdriver.Chrome()
    driver.set_page_load_timeout(10)
    driver.get(row['URL'])
    time.sleep(7)
    reset_button = driver.find_element(By.CSS_SELECTOR, f"#contentsArea > div.pager > a:nth-child({page_number+3})")
    #reset_button = driver.find_element(By.XPATH, f"//*[@id='list']/div[2]/div/a[{page_number+2}]")
    reset_button.click()
    print(f"#contentsArea > div.pager > a:nth-child({page_number+3})")
    time.sleep(7)
    page_source = driver.page_source
    soup = BeautifulSoup(page_source, 'html.parser')
    driver.quit()
    return soup

def click_dynamic_crawl30(row, page_number):
    driver = webdriver.Chrome()
    driver.set_page_load_timeout(10)
    driver.get(row['URL'])
    time.sleep(7)
    reset_button = driver.find_element(By.XPATH, f"/html/body/form/div/table/tbody/tr/td/table/tbody/tr/td/table/tbody/tr/td[4]/span[{(page_number-1)*2+1}]")
    #reset_button = driver.find_element(By.XPATH, f"//*[@id='list']/div[2]/div/a[{page_number+2}]")
    reset_button.click()
    print(f"/html/body/form/div/table/tbody/tr/td/table/tbody/tr/td/table/tbody/tr/td[4]/span[{(page_number-1)*2+1}]")
    time.sleep(7)
    page_source = driver.page_source
    soup = BeautifulSoup(page_source, 'html.parser')
    driver.quit()
    return soup

def click_dynamic_crawl31(row, page_number):
    driver = webdriver.Chrome()
    driver.set_page_load_timeout(10)
    driver.get(row['URL'])
    time.sleep(7)
    reset_button = driver.find_element(By.CSS_SELECTOR, f"body > form > section > div.pager > a:nth-child({(page_number+2)}")
    #reset_button = driver.find_element(By.XPATH, f"//*[@id='list']/div[2]/div/a[{page_number+2}]")
    reset_button.click()
    print(f"body > form > section > div.pager > a:nth-child({(page_number+2)}")
    time.sleep(7)
    page_source = driver.page_source
    soup = BeautifulSoup(page_source, 'html.parser')
    driver.quit()
    return soup

def click_dynamic_crawl32(row, page_number):
    driver = webdriver.Chrome()
    driver.set_page_load_timeout(10)
    driver.get(row['URL'])
    time.sleep(7)
    reset_button = driver.find_element(By.XPATH, f"/html/body/div/main/div/div/div[2]/div[2]/div[3]/a[{(page_number+2)}]")
    #reset_button = driver.find_element(By.XPATH, f"//*[@id='list']/div[2]/div/a[{page_number+2}]")
    reset_button.click()
    print(f"/html/body/div/main/div/div/div[2]/div[2]/div[3]/a[{(page_number+2)}]")
    time.sleep(7)
    page_source = driver.page_source
    soup = BeautifulSoup(page_source, 'html.parser')
    driver.quit()
    return soup

def click_dynamic_crawl33(row, page_number):
    driver = webdriver.Chrome(options=chrome_options)
    driver.set_page_load_timeout(10)
    driver.get(row['URL'])
    time.sleep(7)
    reset_button = driver.find_element(By.CSS_SELECTOR, f"body > form > table:nth-child(12) > tbody > tr > td > table:nth-child(3) > tbody > tr > td > table > tbody > tr > td > table > tbody > tr > td:nth-child(4) > span[{(page_number-1)*2+1}]/a")
    reset_button.click()
    print(f"body > form > table:nth-child(12) > tbody > tr > td > table:nth-child(3) > tbody > tr > td > table > tbody > tr > td > table > tbody > tr > td:nth-child(4) > span[{(page_number-1)*2+1}]/a")
    time.sleep(7)
    page_source = driver.page_source
    soup = BeautifulSoup(page_source, 'html.parser')
    driver.quit()
    return soup

def click_dynamic_crawl34(row, page_number):
    driver = webdriver.Chrome(options=chrome_options)
    driver.set_page_load_timeout(10)
    driver.get(row['URL'])
    time.sleep(7)
    reset_button = driver.find_element(By.CSS_SELECTOR, f"#paging-tag > ul > li:nth-child({page_number+2})")
    reset_button.click()
    print(f"#paging-tag > ul > li:nth-child({page_number+2})")
    time.sleep(7)
    page_source = driver.page_source
    soup = BeautifulSoup(page_source, 'html.parser')
    driver.quit()
    return soup

# 정적 크롤링
def static_crawl(row):
    soup = None  # 기본값으로 soup을 None으로 초기화
    try:
      response = requests.get(row['URL'], headers=headers, timeout=(50, 50), verify=False)
      response.raise_for_status()  # 응답이 200번대가 아니면 예외 발생
      response.encoding = 'utf-8'
      soup = BeautifulSoup(response.text, 'html.parser')
    except requests.exceptions.ConnectTimeout:
      print(f"연결 타임아웃: {row['SITE_NAME']} 서버로부터 응답이 없습니다.")
    except requests.exceptions.ReadTimeout:
      print(f"읽기 타임아웃: {row['SITE_NAME']} 서버가 데이터를 제공하는 시간이 초과되었습니다.")
    except requests.exceptions.RequestException as e:
      print(f"요청 오류: {e}")
    return soup

# post 크롤링
def post_crawl(row):
    data = {
    'epcCheck': '',
    'pageIndex': '',
    'jndinm': 'OfrNotAncmtEJB',
    'context': 'NTIS',
    'method': 'selectListOfrNotAncmt',
    'methodnm': 'selectListOfrNotAncmtHomepage',
    'not_ancmt_mgt_no': '',
    'homepage_pbs_yn': 'Y',
    'subCheck': 'N',
    'ofr_pageSize': '10',
    'not_ancmt_se_code': '01,04,06',
    'title': '고시공고',
    'cha_dep_code_nm': '',
    'initValue': '',
    'countYn': 'Y',
    'list_gubun': 'A',
    'not_ancmt_sj': '',
    'not_ancmt_cn': '',
    'dept_nm': '',
    'cgg_code': '',
    'yyyy': '',
    'yyyymmdd': '',
    'recent_mm': '',
    'last_mm': '',
    'nodate_recent_mm': '',
    'nodate_last_mm': '',
    'not_ancmt_reg_no': '',
    'Key': 'B_Subject',
    'temp': ''
    }
    response = requests.post(row['URL'], data=data).content
    page_source = response.decode('utf-8-sig')
    soup = BeautifulSoup(page_source, 'html.parser')
    return soup

# '작성일' 형식 수정 함수
def fix_date_format(date_str):
    # YYYYMMDD 형식을 YYYY-MM-DD로 변환
    if len(date_str) == 8 and date_str.isdigit():
        return f"{date_str[:4]}-{date_str[4:6]}-{date_str[6:]}"

    if len(date_str) == 18:
      return f"{date_str[:4]}-{date_str[5:7]}-{date_str[8:10]}"

    # 날짜 범위가 있을 경우 첫 번째 날짜만 사용
    date_str = date_str.split('~')[0].strip()

    # 2자리 연도를 4자리로 변환
    parts = date_str.split('-')
    if len(parts[0]) == 2:  # 연도가 2자리인 경우
        return '20' + date_str

    return date_str

def update_url_for_next_page(url, page_number, div):
    # URL의 pageIndex 파라미터를 증가시키기 위해 함수 정의
    if page_number == 1:
        return url  # page_number가 1일 때는 원래 URL 그대로 반환
    elif div == 'V2' and "pageIndex=" in url:
        return re.sub(r"pageIndex=\d+", f"pageIndex={page_number}", url)
    elif div == 'V2' and "page=" in url:
        base_url, current_page = url.split("page=")
        return f"{base_url}page={page_number}"
    elif div == 'V2' and "Page=" in url:
        base_url, current_page = url.split("Page=")
        return f"{base_url}Page={page_number}"
    elif div == 'V2' and "&cpn=" in url:
        base_url, current_page = url.split("&cpn=")
        return f"{base_url}&cpn={page_number}"
    elif div == 'V2' and "pageNo=" in url:
        base_url, current_page = url.split("pageNo=")
        return f"{base_url}pageNo={page_number}"
    elif div == 'V2' and "offset=" in url:
        base_url, current_page = url.split("offset=")
        return f"{base_url}offset={(page_number-1)*15}"
    elif div == 'V2' and "?p=" in url:
        base_url, current_page = url.split("?p=")
        return f"{base_url}?p={page_number}"
    elif div == 'V2' and "Page2=" in url:
        base_url, current_page = url.split("Page2=")
        return f"{base_url}Page2={page_number}"
    elif div == 'V2' and "Start=" in url:
        base_url, current_page = url.split("Start=")
        return f"{base_url}Start={(page_number-1)*10}"
    elif div == 'V2' and "pageid=" in url:
        base_url, current_page = url.split("pageid=")
        return f"{base_url}pageid={page_number}"
    elif div == 'V2':
        return url
    else :
        print("URL pattern is developing. Stopping.")
        return None

def update_crawl_type(url, crawl_type, page_number, ct2):
  if page_number == 1:
    return crawl_type
  elif not (isinstance(ct2, float) and np.isnan(ct2)):  # NaN 체크
      return ct2
  else:
      return crawl_type
```

```python
def extract_detail_url(a_tag, url_cd):
    """
    각 사이트 구조별로 <a> 태그에서 상세 URL을 추출하는 함수
    """

    print("a_tag", a_tag)
    href = a_tag.get("href", "")

    if url_cd == "jeonnam":  # 전라남도
        base_url = "https://www.jeonnam.go.kr"
        return base_url + href

    elif url_cd == "gangwon":  # 강원도(goPage() 형태)
        match = re.search(r"goPage\((\d+)\)", a_tag.get("onclick", ""))
        if match:
            article_seq = match.group(1)
            return f"https://state.gwd.go.kr/road/announce?pageIndex=1&recordCountPerPage=90&mode=&firstYN=N&articleSeq={article_seq}"
        return None

    elif url_cd == "chungnam":
        return "https://www.chungnam.go.kr" + href

    elif url_cd == "daejeon":  # 대전광역시 등 일반 상대 경로
        return requests.compat.urljoin("https://www.daejeon.go.kr/", href)

    elif url_cd == "yeoju":  # ✅ 여주시 상대경로 처리
        return requests.compat.urljoin("https://www.yeoju.go.kr/www/", href)

    elif url_cd == "mokpo":
      return requests.compat.urljoin("https://www.mokpo.go.kr/", href)

    elif url_cd == "chungbuk":
      return requests.compat.urljoin("https://www.chungbuk.go.kr/www/", href)

    # 추가 케이스 정의 가능
    else:
        return None  # 기본은 None 반환
```

```python
# InsecureRequestWarning 경고 메시지 억제
requests.packages.urllib3.disable_warnings(InsecureRequestWarning)

headers = {"User-Agent" : "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/127.0.0.0 Safari/537.36",
      "Accept-Language" : "ko-KR,ko;q=0.9,en-US;q=0.8,en;q=0.7"
}

# 크롤링 시작
for index, row in tqdm(df.iterrows(), total=df.shape[0], desc="Processing rows"):
    retries = 0
    page_number = 1
    cleaned_dates = []  # 누적된 날짜 저장용 리스트
    all_titles = []  # 누적된 제목 저장용 리스트
    max_pages = 9  # 무한 루프 방지: 최대 페이지 시도 제한

    while page_number <= max_pages:
        try:
          # URL을 페이지 번호에 따라 업데이트
            updated_url = update_url_for_next_page(row['URL'], page_number, row['div'])
            if updated_url is None:
                print(f"Unable to update URL for pagination on {row['SITE_NAME']}. Stopping pagination.")
                break  # URL을 업데이트할 수 없는 경우 루프 중단
            row['URL'] = updated_url
            row['crawl_type'] = update_crawl_type(row['URL'], row['crawl_type'], page_number, row['ct2'])
            print(f"Updated URL for {row['SITE_NAME']} to: {row['URL']}")  # URL 상태 출력
            success = False  # 페이지별 크롤링 성공 여부 초기화

            print(f"Processing {row['SITE_NAME']} - Page: {page_number}")  # 상태 체크를 위한 로그

            while retries < MAX_RETRIES and not success:
                try:
                    if row['crawl_type'] == 's':
                        soup = static_crawl(row)  # 정적 크롤링 수행
                    elif row['crawl_type'] == 'd':
                        soup = dynamic_crawl(row)  # 동적 크롤링 수행
                    elif row['crawl_type'] == 'd1':
                        soup = dynamic_crawl_1(row)  # 동적 크롤링 수행
                    elif row['crawl_type'] == 'd2':
                        soup = dynamic_crawl_2(row)  # 동적 크롤링 수행
                    elif row['crawl_type'] == 'p':
                        soup = post_crawl(row)  # post 크롤링 수행
                    elif row['crawl_type'] == 'cd':
                        soup = click_dynamic_crawl(row, page_number)
                    elif row['crawl_type'] == 'cd1':
                        soup = click_dynamic_crawl1(row, page_number)
                    elif row['crawl_type'] == 'cd2':
                        soup = click_dynamic_crawl2(row, page_number)
                    elif row['crawl_type'] == 'cd3':
                        soup = click_dynamic_crawl3(row, page_number)
                    elif row['crawl_type'] == 'cd4':
                        soup = click_dynamic_crawl4(row, page_number)
                    elif row['crawl_type'] == 'cd5':
                        soup = click_dynamic_crawl5(row, page_number)
                    elif row['crawl_type'] == 'cd6':
                        soup = click_dynamic_crawl6(row, page_number)
                    elif row['crawl_type'] == 'cd7':
                        soup = click_dynamic_crawl7(row, page_number)
                    elif row['crawl_type'] == 'cd8':
                        soup = click_dynamic_crawl8(row, page_number)
                    elif row['crawl_type'] == 'cd9':
                        soup = click_dynamic_crawl9(row, page_number)
                    elif row['crawl_type'] == 'cd10':
                        soup = click_dynamic_crawl10(row, page_number)
                    elif row['crawl_type'] == 'cd11':
                        soup = click_dynamic_crawl11(row, page_number)
                    elif row['crawl_type'] == 'cd12':
                        soup = click_dynamic_crawl12(row, page_number)
                    elif row['crawl_type'] == 'cd13':
                        soup = click_dynamic_crawl13(row, page_number)
                    elif row['crawl_type'] == 'cd14':
                        soup = click_dynamic_crawl14(row, page_number)
                    elif row['crawl_type'] == 'cd15':
                        soup = click_dynamic_crawl15(row, page_number)
                    elif row['crawl_type'] == 'cd16':
                        soup = click_dynamic_crawl16(row, page_number)
                    elif row['crawl_type'] == 'cd17':
                        soup = click_dynamic_crawl17(row, page_number)
                    elif row['crawl_type'] == 'cd18':
                        soup = click_dynamic_crawl18(row, page_number)
                    elif row['crawl_type'] == 'cd19':
                        soup = click_dynamic_crawl19(row, page_number)
                    elif row['crawl_type'] == 'cd20':
                        soup = click_dynamic_crawl20(row, page_number)
                    elif row['crawl_type'] == 'cd21':
                        soup = click_dynamic_crawl21(row, page_number)
                    elif row['crawl_type'] == 'cd22':
                        soup = click_dynamic_crawl22(row, page_number)
                    elif row['crawl_type'] == 'cd23':
                        soup = click_dynamic_crawl23(row, page_number) 
                    elif row['crawl_type'] == 'cd24':
                        soup = click_dynamic_crawl24(row, page_number)
                    elif row['crawl_type'] == 'cd25':
                        soup = click_dynamic_crawl25(row, page_number)
                    elif row['crawl_type'] == 'cd26':
                        soup = click_dynamic_crawl26(row, page_number)
                    elif row['crawl_type'] == 'cd27':
                        soup = click_dynamic_crawl27(row, page_number)
                    elif row['crawl_type'] == 'cd28':
                        soup = click_dynamic_crawl28(row, page_number)
                    elif row['crawl_type'] == 'cd29':
                        soup = click_dynamic_crawl29(row, page_number)
                    elif row['crawl_type'] == 'cd30':
                        soup = click_dynamic_crawl30(row, page_number)
                    elif row['crawl_type'] == 'cd31':
                        soup = click_dynamic_crawl31(row, page_number)
                    elif row['crawl_type'] == 'cd32':
                        soup = click_dynamic_crawl32(row, page_number)
                    elif row['crawl_type'] == 'cd33':
                        soup = click_dynamic_crawl33(row, page_number)
                    elif row['crawl_type'] == 'cd34':
                        soup = click_dynamic_crawl34(row, page_number)

                    # 크롤링 성공 시 루프 종료
                    success = True
                    print(f"Successfully crawled {row['SITE_NAME']} - Page: {page_number}")

                except TimeoutException:
                    retries += 1
                    print(f"Timeout occurred for {row['SITE_NAME']} on page {page_number}. Retrying {retries}/{MAX_RETRIES}...")
                    time.sleep(2)

            if not success:
                print(f"Failed to crawl {row['SITE_NAME']} on page {page_number} after {MAX_RETRIES} retries.")
                break

            # 데이터 처리 및 저장 로직
            data = []
            if row['SITE_NAME'] == '대전광역시고시공고':
                table_body = soup.select(row['table_body'])[1]
            else:
                table_body = soup.select_one(row['table_body'])

            titles = table_body.select(row['title'])
            dates = table_body.select(row['date'])

            # 특정 사이트 '충청도_서천군'에서 '등록일' 제외한 날짜만 가져오기
            if row['SITE_NAME'] == '충청도_서천군':
                dates = [date for date in dates if '등록일' not in date.get_text(strip=True)]


            # 데이터 저장
            for title, date in zip(titles, dates):
                clean_title = title.get_text(strip=True).replace("\r", "").replace("\n", "").replace("\t", "").strip()
                text = date.get_text(separator=" ", strip=True)

                if '공고부서 :' in text:
                    extracted_date = text.split('공고부서 :')[-2].split('등록일 :')[-1].strip()
                elif '게재일 :' in text:
                    extracted_date = text.split('게재일 :')[1].strip()
                else:
                    extracted_date = text.replace('.', '-').replace('/', '-').replace('등록일 :', '').strip()

                # 모든 날짜 저장, 제목과 상관없이
                extracted_date = fix_date_format(extracted_date)

                # 상세 URL 추출
                detail_url = extract_detail_url(title, row.get("url_cd",""))
                


                if any(keyword in clean_title for keyword in ['특허', '제안', '심의', '공법','실시설계','보수보강']):
                    data.append({
                        "SITE_NO": row['SITE_NO'],
                        "출처": row['SITE_NAME'],
                        "URL": detail_url if detail_url else row['URL'],
                        "URL": row['URL'],
                        "제목": clean_title,
                        "작성일": extracted_date
                    })
                all_titles.append(clean_title)
                cleaned_dates.append(extracted_date)  # Append all extracted dates, regardless of filtering

            tmp_df = pd.DataFrame(data)
            try:
                tmp_df['작성일'] = tmp_df['작성일'].apply(fix_date_format)
            except KeyError:
                pass
            tmp_df = tmp_df.drop_duplicates()

            # DataFrame에 결과 추가
            if 'df_fin' not in locals():
                df_fin = tmp_df
            else:
                df_fin = pd.concat([df_fin, tmp_df], ignore_index=True)

            # 페이지 이동 조건 체크 (전체 누적된 날짜 기반)
            cumulative_unique_dates = len(set(cleaned_dates))
            if cumulative_unique_dates <= 2:
                page_number += 1  # 페이지 넘버 증가
                retries = 0  # 재시도 횟수 초기화
                time.sleep(1)
                print(f"Cumulative unique dates so far: {cumulative_unique_dates}")
            else:
                print(f"Stopping {row['SITE_NAME']} - sufficient data collected.")
                break

        except Exception as e:
            print(f"Error occurred during processing {row['SITE_NAME']} on page {page_number}: {e}")
            break

    # min_date, max_date, unique_date 계산
    if cleaned_dates:
        try:
            min_date = min(cleaned_dates)
            max_date = max(cleaned_dates)
            unique_date = len(set(cleaned_dates))
        except ValueError:
            min_date = None
            max_date = None
            unique_date = 0
    else:
        min_date = None
        max_date = None
        unique_date = 0

    # 로그 업데이트
    df.at[index, 'len_tbody'] = len(all_titles)
    df.at[index, 'unique_date'] = unique_date
    df.at[index, 'min_date'] = min_date if min_date else ""
    df.at[index, 'max_date'] = max_date if max_date else ""
```
